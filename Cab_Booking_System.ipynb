# **Uber/Ola-Style Cab Booking System**




```python
#PARENT CLASS - VEHICLE

class Vehicle:
    def __init__(self, vehicle_number, brand, driver_name, price_per_km):
        self.vehicle_number = vehicle_number
        self.brand = brand
        self.driver_name = driver_name
        self.price_per_km = price_per_km

    def calculate_fare(self, distance):
        return distance * self.price_per_km

    def display_details(self):
        print(f"Vehicle Number : {self.vehicle_number}")
        print(f"Brand          : {self.brand}")
        print(f"Driver         : {self.driver_name}")
        print(f"Price per KM   : ₹{self.price_per_km}")
```

```python
#2. CHILD CLASS - CAR

class Car(Vehicle):
    def __init__(self, vehicle_number, brand, driver_name, price_per_km):
        super().__init__(vehicle_number, brand, driver_name, price_per_km)
        self.vehicle_type = "Car"
```

```python
#3. CHILD CLASS - BIKE

class Bike(Vehicle):
    def __init__(self, vehicle_number, brand, driver_name, price_per_km):
        super().__init__(vehicle_number, brand, driver_name, price_per_km)
        self.vehicle_type = "Bike"
```

```python
# Creating 2 Cars

car1 = Car("KL01AB1234", "Toyota", "Anu", 20)
car2 = Car("KL02CD5678", "Honda", "Manu", 18)
```

```python
# Creating 2 Bikes

bike1 = Bike("KL03EF1111", "Honda", "Binu", 10)
bike2 = Bike("KL04GH2222", "Yamaha", "Meena", 12)
```

```python
# Multiple trips
trips = [
    (car1, 15),
    (car2, 20),
    (bike1, 10),
    (bike2, 25)
]
```

```python
# Display trip details and fare
for vehicle, distance in trips:
    print("\n-----------------------------")
    print(f"Vehicle Type   : {vehicle.vehicle_type}")
    print(f"Driver         : {vehicle.driver_name}")
    print(f"Vehicle Number : {vehicle.vehicle_number}")
    print(f"Brand          : {vehicle.brand}")
    print(f"Distance       : {distance} KM")
    print(f"Rate           : ₹{vehicle.price_per_km}/KM")

    fare = vehicle.calculate_fare(distance)
    print(f"Total Fare     : ₹{fare}")
```
